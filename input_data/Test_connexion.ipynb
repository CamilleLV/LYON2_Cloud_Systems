## Test de connexion HDFS

```python
from pyspark.sql import SparkSession

# 1. Initialisation de la session Spark
print("Démarrage de Spark...")
spark = SparkSession.builder \
    .appName("Test-Hadoop") \
    .master("local[*]") \
    .config("spark.hadoop.fs.defaultFS", "hdfs://namenode:9000") \
    .getOrCreate()

print("Spark est prêt !")

# 2. Création de données de test (en mémoire)
data = [("Alice", 25), ("Bob", 30), ("Charlie", 35)]
columns = ["Nom", "Age"]
df = spark.createDataFrame(data, columns)

print("--- Données créées en mémoire ---")
df.show()

# 3. Écriture dans HDFS (Le système de fichiers Hadoop)
# On écrit dans le dossier /user/test_write
hdfs_path = "hdfs://namenode:9000/user/test_write"
print(f"Écriture des données vers : {hdfs_path} ...")

# Le mode 'overwrite' écrase le dossier s'il existe déjà
df.write.mode("overwrite").csv(hdfs_path)
print("Écriture terminée !")

# 4. Lecture depuis HDFS pour vérifier
print("Lecture des données depuis Hadoop...")
df_read = spark.read.csv(hdfs_path)
df_read.show()

print("Succès ! La connexion Spark <-> Hadoop fonctionne.")
```
